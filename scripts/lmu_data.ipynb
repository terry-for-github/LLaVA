```python
import pandas as pd
import json
import os

lmu_home = '/userhome/Dataset/LMUData/'
datasets = ['AI2D_TEST', 'HallusionBench', 'MMStar', 'MME', 'RealWorldQA', 'OCRBench']
list_data_dict = []
cnt = 0
i_cnt = 0
for dataset_idx, dataset in enumerate(datasets):
    dataset_path = lmu_home + dataset + '.tsv'
    data = pd.read_csv(dataset_path, sep='\t')
    print(dataset, len(data))
    for idx in range(len(data)):
        prompt, answer, image_path = '', '', ''
        if dataset in ['AI2D_TEST', 'MMStar', 'RealWorldQA', 'MMBench_DEV_EN_V11']:
            prompt = '<image>\n' + data['question'][idx]
            if isinstance(data['A'][idx], str):
                prompt += '\nA. ' + data['A'][idx]
            if isinstance(data['B'][idx], str):
                prompt += '\nB. ' + data['B'][idx]
            if isinstance(data['C'][idx], str):
                prompt += '\nC. ' + data['C'][idx]
            if isinstance(data['D'][idx], str):
                prompt += '\nD. ' + data['D'][idx]
            prompt += "\nAnswer with the option's letter from the given choices directly."
            answer = data['answer'][idx]
        elif dataset in ['HallusionBench']:
            prompt = '<image>\n' +  data['question'][idx]
            answer = data['answer'][idx] + '. ' + data['gt_answer_details'][idx]
        elif dataset in ['MME']:
            prompt = '<image>\n' + data['question'][idx]
            answer = data['answer'][idx]
        elif dataset in ['OCRBench']:
            prompt = '<image>\n' + data['question'][idx]
            answer = data['answer'][idx][2:-2]
        dataset_name = dataset if dataset != 'MMBench_DEV_EN_V11' else 'MMBench_V11'
        image_path = dataset_name + '/' + str(data['index'][idx]) + '.jpg'
        if not os.path.exists(os.path.join(lmu_home, 'images', image_path)):
            i_cnt += 1
            continue
        # if dataset_idx == 3:
        #     print(prompt, answer, image_path, sep='\n')
        # 
        data_dict = {
            'id': cnt,
            'conversations': [
                {
                    'from': 'human',
                    'value': prompt
                },
                {
                    'from': 'gpt',
                    'value': answer
                }
            ],
            'image': image_path
        }
        list_data_dict.append(data_dict)
        cnt += 1
print(cnt)
print(i_cnt)
json.dump(list_data_dict, open('lmu_data.json', 'w'), indent=4)
```
